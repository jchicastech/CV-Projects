# Image Processing Basics

This notebook introduces fundamental image processing operations using **OpenCV**, **NumPy**, and **Matplotlib**. Topics covered:

1. Loading and displaying images
2. Color spaces (BGR, RGB, Grayscale, HSV)
3. Basic geometric transformations
4. Image filtering (blur, sharpen)
5. Edge detection (Canny)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Create a Sample Image

We generate a simple synthetic image so no external dataset is required.

```python
# Create a 256x256 RGB test image with geometric shapes
img = np.zeros((256, 256, 3), dtype=np.uint8)

# Blue filled rectangle
cv2.rectangle(img, (20, 20), (120, 120), (255, 0, 0), -1)

# Green filled circle
cv2.circle(img, (190, 80), 55, (0, 255, 0), -1)

# Red filled triangle (polygon)
triangle = np.array([[128, 200], [50, 240], [206, 240]], dtype=np.int32)
cv2.fillPoly(img, [triangle], (0, 0, 255))

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Sample Image (BGR → RGB for display)')
plt.axis('off')
plt.show()
```

## 2. Color Spaces

OpenCV loads images in **BGR** order. We convert to common color spaces for analysis.

```python
rgb   = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray  = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv   = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
titles = ['Original (BGR→RGB)', 'Grayscale', 'HSV – Hue', 'HSV – Saturation']
images = [rgb, gray, hsv[:, :, 0], hsv[:, :, 1]]
cmaps  = [None, 'gray', 'hsv', 'gray']

for ax, title, image, cmap in zip(axes, titles, images, cmaps):
    ax.imshow(image, cmap=cmap)
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 3. Geometric Transformations

Resize, rotate, and flip operations are common preprocessing steps in CV pipelines.

```python
# Resize to half
resized = cv2.resize(img, (128, 128))

# Rotate 45 degrees around the center
h, w = img.shape[:2]
M = cv2.getRotationMatrix2D((w / 2, h / 2), 45, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))

# Horizontal flip
flipped = cv2.flip(img, 1)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, title, image in zip(
    axes,
    ['Resized (128×128)', 'Rotated 45°', 'Flipped (horizontal)'],
    [resized, rotated, flipped]
):
    ax.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 4. Image Filtering

Filtering smooths noise or enhances fine detail.

```python
# Add Gaussian noise for demonstration
noise = np.random.normal(0, 25, img.shape).astype(np.int16)
noisy = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8)

# Gaussian blur (smoothing)
blurred = cv2.GaussianBlur(noisy, (5, 5), sigmaX=1.5)

# Sharpening kernel
kernel_sharpen = np.array([[ 0, -1,  0],
                            [-1,  5, -1],
                            [ 0, -1,  0]], dtype=np.float32)
sharpened = cv2.filter2D(img, -1, kernel_sharpen)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, title, image in zip(
    axes,
    ['Noisy', 'Gaussian Blur', 'Sharpened'],
    [noisy, blurred, sharpened]
):
    ax.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 5. Edge Detection (Canny)

The Canny algorithm is one of the most widely used edge detectors.

```python
edges_low  = cv2.Canny(gray, threshold1=30,  threshold2=100)
edges_mid  = cv2.Canny(gray, threshold1=80,  threshold2=160)
edges_high = cv2.Canny(gray, threshold1=150, threshold2=250)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, title, edges in zip(
    axes,
    ['Canny (low thresholds)', 'Canny (mid thresholds)', 'Canny (high thresholds)'],
    [edges_low, edges_mid, edges_high]
):
    ax.imshow(edges, cmap='gray')
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## Summary

| Technique | OpenCV function |

| Flip | `cv2.flip` |
| Gaussian blur | `cv2.GaussianBlur` |
